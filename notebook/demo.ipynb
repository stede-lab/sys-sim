# Initialize

```python
%reload_ext autoreload
%autoreload 2
import sys
sys.path.append('../src')
import syside
import ast
from transpile import (transformAction)
```

# Load Model

```python
MODEL_FILES = [
  "/home/sherburne/ubuntu-24-041/sys-sim/sysml/simple-action.sysml"
]

ast_funcs = []

(model, diagnostics) = syside.load_model(MODEL_FILES)
assert list(diagnostics.all) == [], "No warnings and no informational messages."

for action in model.nodes(syside.ActionDefinition):
  # Create Function
  ast_funcs.append(transformAction(action))

# Create the AST
ast_tree = ast.Module(ast_funcs, type_ignores=[])
ast.fix_missing_locations(ast_tree)

# Print the AST in a readable format
print("--------------------")
print(ast.dump(ast_tree, indent=2))
print("--------------------")
print(ast.unparse(ast_tree))

# Execute the AST
exec(compile(ast_tree, filename="<ast>", mode="exec"))
 
```
